## 1. Sambungkan Google Drive

## 2. Unduh Flickr8k dari Kaggle

Sumber: https://www.kaggle.com/datasets/adityajn105/flickr8k

```python
!pip install -q kagglehub
import kagglehub, os

path = kagglehub.dataset_download("adityajn105/flickr8k")
print('Lokasi:', path)
print('Isi   :', os.listdir(path))   # harus ada: Images, captions.txt
print('Jumlah gambar:', len(os.listdir(os.path.join(path, 'Images'))))   # harus 8091
```

## 3. Simpan sebagai `archive.zip` di Drive

```python
import shutil

tujuan = '/content/drive/MyDrive/ImageCaptioning/data'
os.makedirs(tujuan, exist_ok=True)

# Zip dibuat di disk Colab dulu (lebih cepat), lalu disalin ke Drive
shutil.make_archive('/content/archive', 'zip', path)
shutil.copy('/content/archive.zip', f'{tujuan}/archive.zip')
print('Selesai:', round(os.path.getsize(f'{tujuan}/archive.zip') / 1e6), 'MB ->', f'{tujuan}/archive.zip')
```
